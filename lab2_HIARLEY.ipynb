# Medidas de Posição **(Média, Mediana e Moda)**



## O que é uma medida de posição?

Uma **medida de posição** (ou de tendência central) responde a uma única pergunta: *se eu tivesse que resumir toda essa amostra em um número só, qual número eu escolheria?*

São três as clássicas, e elas **não** são intercambiáveis:

| Medida | Definição | Quando usar |
|---|---|---|
| **Média** | Soma dos valores dividida por *n* | Dados simétricos, sem valores extremos |
| **Mediana** | O valor central da amostra ordenada | Quando há valores extremos (renda, preço de imóvel) |
| **Moda** | O valor que mais se repete | Variáveis qualitativas, ou para achar o "padrão" |

```python

# --------------------------------------------------------------------
# MÉDIA ARITMÉTICA
# Fórmula: media = (soma de todos os valores) / (quantidade de valores)
# --------------------------------------------------------------------

def media(amostra):
    n = len(amostra)
    if n == 0:
      return None
    return somatorio(amostra) / n

# Aplicando nossa amostra de 7 lanches:
media_kcal = media(kcal_amostra)

print("Amostra:", kcal_amostra)
print("Soma   :", somatorio(kcal_amostra))
print("n      :", len(kcal_amostra))
print(f"Média  :{media_kcal:.2f} kcal")
```

### A mediana

$$
\tilde{x} =
\begin{cases}
x_{\left(\frac{n+1}{2}\right)} & \text{se } n \text{ é ímpar} \\[6pt]
\dfrac{x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2}+1\right)}}{2} & \text{se } n \text{ é par}
\end{cases}
$$

O algoritmo tem **três passos** e o primeiro é o que todo mundo esquece:

1. **Ordenar** a amostra (sem isso, não existe "valor central");
2. Se *n* for **ímpar**, pegar o valor do meio;
3. Se *n* for **par**, tirar a média dos **dois** valores centrais.

> **Atenção ao índice:** a posição $\frac{n+1}{2}$ da fórmula estatística conta a partir de **1**. O Python conta a partir de **0**. Por isso, no código, subtraímos 1 da posição.

```python
# --------------------------------------------------------------
# Calculando a Mediana à Mão
# --------------------------------------------------------------

def mediana(amostra):
  ordenada = sorted(amostra)
  n = len(ordenada)

  if n % 2 == 1:
    meio = n // 2
    return ordenada[meio]

  else:
    direita = n // 2
    esquerda = direita - 1
    return (ordenada[esquerda] + ordenada[direita]) / 2

print("Amostra Original :", kcal_amostra)
print("Amostra Ordenada :", sorted(kcal_amostra))
print("n                :", len(kcal_amostra), "(ímpar)" if len(kcal_amostra) % 2 else "(ímpar)")
print(f"Média           : {media(kcal_amostra):.2f} kcal")
print(f"Mediana         : {mediana(kcal_amostra):.2f} kcal")
print()
print("Repare que média e mediana NÃO são iguais. A diferença entre elas")
print("é o primeiro sinal de que a distribuição é assimétrica.")


```

### A moda

A **moda** é o valor que aparece com maior frequência na amostra.

É a única medida de posição que funciona em variável **qualitativa**: não existe "média de categoria de lanche", mas existe perfeitamente "a categoria mais frequente é Frango".

O algoritmo em duas etapas:
1. Montar a **tabela de frequências** (um dicionário valor → contagem);
2. Encontrar a chave com a maior contagem.

Uma amostra pode ser **amodal** (nenhum valor se repete), **unimodal** (uma moda) ou **multimodal** (empate). No nosso caso de kcal, todos os valores são distintos — o que já é uma informação relevante.

```python
# ---------------------------------------------------------------
# MODA - Implementada na mão, vida tabela de frequência
# ---------------------------------------------------------------

# PASSO 1: Vamos montar uma tabela de frequência absoluta de uma amostra.
def tabela_frequencias(amostra):

  frequencias = {}
  for valor in amostra:
      frequencias[valor] = frequencias.get(valor, 0) + 1
  return frequencias


def moda(amostra):

  frequencias = tabela_frequencias(amostra)
  maior = max(frequencias.values())

  modas = []
  for valor, contagem in frequencias.items():
    if contagem == maior:
        modas.append(valor)
  return modas


# TESTE 01 - variável quantitativa sem repetição (Amodal na prática)
print("Frequências das Kcal   :", tabela_frequencias(kcal_amostra))
print("Moda das kcal          :", moda(kcal_amostra), "Todos aparecem uma vez, sem moda útil")

# TESTE 02 - Variável QUALITATIVA, onde a moda é a única medida possível.
categorias = ["Bovino", "Frango", "Bovino", "Peixe", "Frango", "Frango", "Bovino"]
print("Categorias:", categorias)
print("Frequência de categorias: ", tabela_frequencias(categorias))
print("Moda das categorias:", moda(categorias))
```

# Etapa 4 — Medidas de dispersão, na mão

## Por que a média não basta

Dois cardápios podem ter **exatamente a mesma média** de 494 kcal e serem completamente diferentes:

- **Cardápio A:** todos os lanches entre 470 e 520 kcal → o cliente sabe o que esperar.
- **Cardápio B:** metade com 260 kcal, metade com 730 kcal → a média de 494 descreve um lanche que **não existe no cardápio**.

A média sozinha esconde isso. A **dispersão** é a medida que revela.

| Medida | Fórmula | O que responde |
|---|---|---|
| **Amplitude** | $x_{max} - x_{min}$ | Qual a distância entre o extremo mínimo e o máximo? |
| **Variância** | $s^2 = \dfrac{1}{n-1}\sum (x_i - \bar{x})^2$ | Quanto os dados se espalham (em unidade ao quadrado)? |
| **Desvio-padrão** | $s = \sqrt{s^2}$ | O mesmo, **na unidade original** |
| **Coef. de variação** | $CV = \dfrac{s}{\bar{x}} \times 100$ | Quanto isso representa em % da média? |

```python
# DEMONSTRAÇÃO — por que não se pode somar os desvios sem elevar ao quadrado.
#
# Calculamos o desvio de cada valor em relação à média e somamos.
# Resultado esperado: praticamente zero (a diferença mínima que aparece
# é apenas erro de arredondamento de ponto flutuante do computador).

print(f"{'Lanche:':<14} {'Valor:':>6} {'Desvio:':>9} {'Desvio²':>12}")
print("-" * 45)

media_amostra = media(kcal_amostra)

soma_desvios = 0
soma_desvios_quadrado = 0

for nome, valor in zip(nomes_amostra, kcal_amostra):
  desvio = valor - media_amostra
  soma_desvios += desvio
  soma_desvios_quadrado += desvio ** 2

  print(f"{nome:<14} {valor:>6} {desvio:>9.2f} {desvio ** 2:>12.2f}")

print("-" * 45)
print(f"{'SOMA:':<14} {'':>6} {soma_desvios:>9.2f} {soma_desvios_quadrado:>12.2f}")
```

### Variância: por que dividir por *n − 1*?

Esta é a dúvida que mais aparece em prova, então vale a explicação completa.

- Se os seus dados são **toda a população** (ex.: as notas de **todos** os 40 alunos da turma), divide-se por **n**. Isso é a variância populacional, $\sigma^2$.
- Se os seus dados são uma **amostra** de algo maior (ex.: 25 lanches de um cardápio que tem centenas de itens), divide-se por **n − 1**. Isso é a variância amostral, $s^2$.

**O motivo:** ao usar a média *da amostra* no lugar da média verdadeira *da população*, a soma dos quadrados fica sistematicamente **menor** do que deveria — a média amostral é, por construção, o ponto que minimiza essa soma. Dividir por um número menor (`n-1` em vez de `n`) compensa esse encolhimento. Chamamos isso de **correção de Bessel**, e o resultado é um estimador **não enviesado**.

> **Na prática:** em Machine Learning quase sempre trabalhamos com amostras. Use `n − 1`. E saiba que **`numpy` usa `n` por padrão** (`ddof=0`) enquanto **`pandas` usa `n − 1`** (`ddof=1`) — uma das inconsistências que mais gera confusão entre as duas bibliotecas.

```python
# ---------------------------------------------------------------------
# VARIÂNCIA E DESVIO-PADRÃO — implementados na mão.
#
# Parâmetro "populacional": permite escolher o divisor.
#   populacional=False -> divide por (n - 1)   [amostral, o padrão]
#   populacional=True  -> divide por n         [população inteira]
# ---------------------------------------------------------------------

def variancia(amostra, populacional=False):
  n = len(amostra)
  if n < 2:
   return None

  m = media(amostra)

  soma_quadrados = 0
  for valor in amostra:
    soma_quadrados += (valor -m) ** 2

  divisor = n if populacional else n - 1
  return soma_quadrados / divisor


def desvio_padrao(amostra, populacional=False):
  return variancia(amostra, populacional) ** 0.5

var_amostral = variancia(kcal_amostra)
var_populacional = variancia(kcal_amostra, populacional=True)

print(f"A soma dos quadrados      : {soma_desvios_quadrado:.2f}")
print(f"n                         : {len(kcal_amostra)}")
print()
print(f"Variancia AMOSTRAL (n -1) : {var_amostral:.2f} kcal²")
print(f"Variancia POPULACIONAL (n): {var_populacional:.2f} kcal²")
print()
print(f"Desvio padrão AMOSTRAL     : {desvio_padrao(kcal_amostra):.2f}")
```

```python
# AMPLITUDE E COEFICINETE DE VARIAÇÃO
# Amplitude: Primeiro valor - Último valor

# Coeficiente de Variação: desvio padrao / media * 100
# Regra de prática de Leitura:
# 1 - até 15%: baixa dispersão;
# 2 - entre 15% e 30%: média dispersão;
# 3 - acima de 30%: alta dispersão.

def amplitude(amostra):
  return max(amostra) - min(amostra)

def coeficiente_variacao(amostra):
  m = media(amostra)
  if m == 0:
    return None
  return (desvio_padrao(amostra) / m) * 100


print("RESUMO COMPLETO DA AMOSTRA DE 7 LANCHEs")
print("=" * 44)
print(f"n                  : {len(kcal_amostra)} kcal")
print(f"Mínimo             : {min(kcal_amostra)} kcal")
print(f"Máximo             : {max(kcal_amostra)} kcal")
print(f"Amplitude          : {amplitude(kcal_amostra)} kcal")
print(f"Média              : {media(kcal_amostra):.2f} kcal")
print(f"Mediana            : {mediana(kcal_amostra):.2f} kcal")
print(f"Desvio-padrão      : {desvio_padrao(kcal_amostra):.2f} kcal")
print(f"CV                 : {coeficiente_variacao(kcal_amostra):.2f} %")
print("=" * 44)

cv = coeficiente_variacao(kcal_amostra)
if cv < 15:
  leitura = "Baixa Dispersão"
elif cv <= 30:
  leitura = "Média Dispersão"
else:
  leitura = "Alta dispersão"

print(f"Leitura do CV: {leitura}")
```

## O boxplot — o gráfico que você precisa saber ler

O **boxplot** (diagrama de caixa) é, provavelmente, o gráfico mais denso em informação de toda a estatística descritiva. Ele mostra **cinco medidas ao mesmo tempo**, mais os valores atípicos.

### Como ele é construído

| Elemento | O que representa |
|---|---|
| Borda inferior da caixa | **Q1** (25% dos dados estão abaixo) |
| Linha dentro da caixa | **Mediana** (Q2) |
| Borda superior da caixa | **Q3** (75% dos dados estão abaixo) |
| Altura da caixa | **IQR** = Q3 − Q1, onde estão os **50% centrais** dos dados |
| Hastes (*whiskers*) | Vão até o valor mais extremo que ainda está **dentro** dos limites |
| Pontos isolados | **Outliers** — valores fora dos limites |

### A regra dos limites

$$\text{limite inferior} = Q1 - 1{,}5 \times IQR \qquad \text{limite superior} = Q3 + 1{,}5 \times IQR$$

Qualquer observação fora desse intervalo é marcada como **outlier**.

```python
# CALCULANDO QUARTIS, ANTES DE MEXER NOS GRÁFICOS:

serie = df["Gorduras Totais (g)"]

# .quantile() recebe a proporção de cada quartil.
q1 = serie.quantile(0.25)
q2 = serie.quantile(0.50)
q3 = serie.quantile(0.75)

iqr = q3 - q1
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print("VARIÁVEL: Gorduras Totais (g)")
print("=" * 46)
print(f"Q1 (25%)          : {q1:.2f} g")
print(f"Q2 (mediana)      : {q2:.2f} g")
print(f"Q3 (75%)          : {q3:.2f} g")
print(f"IQR = Q3 - Q1     : {iqr:.2f} g")
print(f"Limite inferior   : {limite_inferior:.2f} g")
print(f"Limite superior   : {limite_superior:.2f} g")
print("=" * 46)

outliers = df[(serie < limite_inferior) | (serie > limite_superior)]

print()
print(f"Foram encontrados {len(outliers)} outliers:")
outliers[["Lanche", "Valor Energético (kcal)", "Gorduras Totais (g)"]]
```
